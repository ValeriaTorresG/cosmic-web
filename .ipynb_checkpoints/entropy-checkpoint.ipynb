```python
from src.classify_web import astra
from src.make_web_catalog import find_fof_groups, compute_group_properties


def compute_void_radii(df_data, df_rand, void_limit, n_iter):
    """
    Ejecuta n_iter iteraciones de ASTRA, para cada iteración:
      - Muestrea rand (tamaño=len(df_data))
      - Clasifica con astra()
      - Selecciona puntos random con r <= void_limit
      - Triangula esos puntos y agrupa por FoF (grupos >=4 pts)
      - Calcula sigma_r de cada void-group
    Devuelve una lista de arrays de radios (sigma_r) por iteración.
    """
    Np = len(df_data)
    radii_list = []
    for it in range(n_iter):
        print(it)
        # 1) submuestra random
        idx = np.random.choice(len(df_rand), size= Np, replace=False)
        sub_rand = df_rand.iloc[idx].reset_index(drop=True)

        # 2) clasificar con ASTRA
        df_typed, coords, tri_all, is_real = astra(
            df_data.copy(), sub_rand.copy()
        )
        r_vals = df_typed['r'].values
        ran_mask = df_typed['RAN'].values

        # 3) conservar solo random void points
        mask = (ran_mask) & (r_vals <= void_limit)
        pos_rand_void = coords[mask]
        if len(pos_rand_void) < 4:
            radii_list.append(np.array([]))
            continue

        # 4) triangulación sobre random-void
        tri_r = Delaunay(pos_rand_void)
        pairs = set()
        for simplex in tri_r.simplices:
            for i,j in combinations(simplex,2):
                a,b = sorted((int(i), int(j)))
                pairs.add((a,b))
        pairs = np.array(sorted(pairs), dtype=int)

        # 5) FoF y filtrar tamaños >=4
        groups = find_fof_groups(pairs)
        groups = {gid: mem for gid, mem in groups.items() if len(mem) >= 4}

        # 6) propiedades y sigma_r
        props = compute_group_properties(groups, pos_rand_void)
        radii = np.array(props['SIGMA_R'])
        radii_list.append(radii)

    return radii_list


def plot_void_size_function(radii_list, n_bins=50):
    """
    Dibuja la PDF de log10(radio) para cada iteración (una línea).
    """
    # Recolectar logR global para definir bins
    all_logR = np.concatenate([
        np.log10(r) for r in radii_list if r.size>0
    ])
    bin_edges = np.linspace(all_logR.min(), all_logR.max(), n_bins+1)
    bin_centers = 0.5*(bin_edges[:-1] + bin_edges[1:])

    plt.figure(figsize=(6,4))
    for radii in radii_list:
        if radii.size==0: continue
        logR = np.log10(radii)
        hist,_ = np.histogram(logR, bins=bin_edges, density=True)
        plt.plot(bin_centers, hist, alpha=0.3, color='black')

    plt.yscale('log')
    plt.xlabel(r'$\log_{10}(R_{\mathrm{void}}\,/\mathrm{Mpc})$')
    plt.ylabel('PDF')
    plt.title('Void size function (ASTRA, datos)')
    plt.tight_layout()

    # os.makedirs(os.path.dirname(output), exist_ok=True)
    # plt.savefig(output, dpi=300)
    # print(f'Figura guardada en {output}')
    plt.show()
```
